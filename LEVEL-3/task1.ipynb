```python
import pandas as pd

df = pd.read_csv("Dataset.csv")
print("Columns in dataset:", df.columns)
```

```python
price_counts = df['Price range'].value_counts().sort_index()
print("Distribution of Price Ranges:")
print(price_counts)

import matplotlib.pyplot as plt
price_counts.plot(kind='bar', color='orange')
plt.title("Distribution of Price Ranges")
plt.xlabel("Price Range")
plt.ylabel("Number of Restaurants")
plt.show()
```

```python
avg_rating_by_price = df.groupby('Price range')['Aggregate rating'].mean()
print("\nAverage Rating by Price Range:")
print(avg_rating_by_price)

avg_rating_by_price.plot(kind='bar', color='green')
plt.title("Average Rating by Price Range")
plt.xlabel("Price Range")
plt.ylabel("Average Rating")
plt.show()
```

```python
avg_votes_by_price = df.groupby('Price range')['Votes'].mean()
print("\nAverage Votes by Price Range:")
print(avg_votes_by_price)

avg_votes_by_price.plot(kind='bar', color='blue')
plt.title("Average Votes by Price Range")
plt.xlabel("Price Range")
plt.ylabel("Average Votes")
plt.show()
```
