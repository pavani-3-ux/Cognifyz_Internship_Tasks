```python
import pandas as pd

# Load dataset
df = pd.read_csv("Dataset.csv")

# Show columns to confirm
print("Columns in dataset:", df.columns)
```

```python
# Count restaurants per city
city_counts = df['City'].value_counts()

top_city = city_counts.idxmax()
top_city_count = city_counts.max()

print(f"City with the highest number of restaurants: {top_city} ({top_city_count} restaurants)")
```

```python
print(df.columns)
```

```python
# Group by city and calculate average rating
avg_ratings = df.groupby('City')['Aggregate rating'].mean()

print("Average rating for each city:")
print(avg_ratings)
```

```python
best_city = avg_ratings.idxmax()
best_city_rating = avg_ratings.max()

print(f"City with the highest average rating: {best_city} ({best_city_rating:.2f})")
```

```python
city_counts = df['City'].value_counts()

top_city = city_counts.idxmax()
top_city_count = city_counts.max()

print(f"City with the highest number of restaurants: {top_city} ({top_city_count} restaurants)")
```

```python
import matplotlib.pyplot as plt

avg_ratings.sort_values(ascending=False).plot(kind='bar', color='orange')
plt.title("Average Restaurant Ratings by City")
plt.ylabel("Average Rating")
plt.show()
```
