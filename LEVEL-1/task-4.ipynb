```python
import pandas as pd

# Load dataset
df = pd.read_csv("Dataset.csv")

# Show columns to confirm
print("Columns in dataset:", df.columns)
```

```python
# Count restaurants by online delivery option
delivery_counts = df['Has Online delivery'].value_counts()

# Calculate percentages
total_restaurants = len(df)
delivery_percentages = (delivery_counts / total_restaurants) * 100

print("Online Delivery Distribution:")
for option, count in delivery_counts.items():
    percentage = delivery_percentages[option]
    print(f"{option}: {count} restaurants ({percentage:.2f}%)")
```

```python
# Compare average ratings for restaurants with and without online delivery
avg_ratings_delivery = df.groupby('Has Online delivery')['Aggregate rating'].mean()

print("Average Ratings by Online Delivery Option:")
print(avg_ratings_delivery)
```

```python
import matplotlib.pyplot as plt

avg_ratings_delivery.plot(kind='bar', color=['blue','red'])
plt.title("Average Ratings: Online Delivery vs No Delivery")
plt.ylabel("Average Rating")
plt.show()
```
