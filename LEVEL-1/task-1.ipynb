```python
import pandas as pd

# Load dataset
df = pd.read_csv("Dataset.csv")

# Show columns to confirm
print("Columns in dataset:", df.columns)
```

```python
# Assuming the column is named 'Cuisines'
top_cuisines = df['Cuisines'].value_counts().head(3)
total_restaurants = len(df)

print("Top 3 Cuisines:")
for cuisine, count in top_cuisines.items():
    percentage = (count / total_restaurants) * 100
    print(f"{cuisine}: {count} restaurants ({percentage:.2f}%)")
```

```python
import matplotlib.pyplot as plt

top_cuisines.plot(kind='bar', color='skyblue')
plt.title("Top 3 Cuisines")
plt.ylabel("Number of Restaurants")
plt.show()
```
