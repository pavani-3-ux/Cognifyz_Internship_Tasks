```python
import pandas as pd

# Load dataset
df = pd.read_csv("Dataset.csv")

# Show columns to confirm
print("Columns in dataset:", df.columns)
```

```python
# Count restaurants in each price range
price_counts = df['Price range'].value_counts()

# Calculate percentages
total_restaurants = len(df)
price_percentages = (price_counts / total_restaurants) * 100

print("Price Range Distribution:")
for price, count in price_counts.items():
    percentage = price_percentages[price]
    print(f"Price Range {price}: {count} restaurants ({percentage:.2f}%)")
```

```python
import matplotlib.pyplot as plt

price_counts.plot(kind='bar', color='green')
plt.title("Distribution of Price Ranges")
plt.xlabel("Price Range Category")
plt.ylabel("Number of Restaurants")
plt.show()
```

```python
df['Price range'].hist(color='purple', bins=range(1, df['Price range'].max()+2))
plt.title("Histogram of Price Ranges")
plt.xlabel("Price Range")
plt.ylabel("Number of Restaurants")
plt.show()
```
